```algrus
АЛГ Пример
НАЧ
| Объявление переменных
ЦЕЛ а, б, результат;
ВЕЩ вещ_число;
ЛОГ флаг;

| Основная программа
а := 10;
б := 5;
вещ_число := 3.14;
флаг := ИСТИНА;

| Условный оператор
ЕСЛИ а > б ТО
    результат := а + б;
ИНАЧЕ
    результат := а - б;
ВСЕ;

| Цикл while
ПОКА результат < 100 ДЕЛАТЬ
    результат := результат * 2;
КОН_ПОКА;

| Цикл for
ДЛЯ i ОТ 1 ДО 5 ШАГ 1 ДЕЛАТЬ
    ВЫВОД "Итерация: ", i;
КОН_ДЛЯ;

| Ввод и вывод
ВВОД а, б;
ВЫВОД "Результат: ", результат;

КОН.
```
